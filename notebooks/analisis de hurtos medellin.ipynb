## Parseo

Las coordenadas vienen con puntos de miles (`627.623.616` → `6.276236`). La edad usa `-1` como centinela. `Sin dato` no es un valor real: se recodifica a faltante **antes** de imputar.

```python
df_p = df_crudo.copy()
df_p.columns = [c.replace("seguridad.", "") for c in df_p.columns]

def parse_coord(valor, tipo):
    """Convierte una coordenada mal formateada a grados decimales.

    En el CSV las coords vienen con puntos de miles, no decimales:
        lat  "627.623.616"   ->  6.27623616   (Medellín ~ 6.2° N)
        lon  "-7.555.353.312" -> -75.55353312  (Medellín ~ 75.5° W)

    La estrategia es: quitar todo lo que no sea dígito, y luego
    insertar el punto decimal después de 1 cifra (lat) o 2 (lon).
    """
    # Valor ausente de pandas (NaN / None / pd.NA)
    if pd.isna(valor):
        return np.nan

    texto = str(valor).strip()
    # Centinelas textuales: tratarlos como faltante, no como coordenada
    if texto.lower() in {"", "nan", "none", "sin dato"}:
        return np.nan

    # Guardamos el signo ANTES de borrar caracteres no numéricos.
    # En Medellín la longitud siempre es negativa (oeste).
    negativo = texto.startswith("-")

    # "627.623.616" -> "627623616";  "-7.555.353.312" -> "7555353312"
    digitos = "".join(ch for ch in texto if ch.isdigit())
    if not digitos:
        return np.nan

    n = int(digitos)

    # Latitud de Colombia: 1 dígito entero (6.xx).
    # Longitud de Colombia: 2 dígitos enteros (75.xx).
    enteros = 1 if tipo == "lat" else 2

    # Dividir por 10^(cantidad de decimales) equivale a insertar el punto.
    # Ej. 9 dígitos y 1 entero -> denom = 10**8 -> 627623616 / 1e8 = 6.27623616
    denom = 10 ** max(len(digitos) - enteros, 0)
    resultado = n / denom if denom else float(n)

    return -resultado if negativo else resultado


def recodificar_centinela(serie):
    texto = serie.astype("string").str.strip()
    mascara = texto.str.casefold().isin(["sin dato", "nan", "none", ""])
    return texto.mask(mascara, pd.NA)


na_antes = df_p.isna().sum()

df_p["fecha"] = pd.to_datetime(df_p["fecha_hecho"], dayfirst=True, errors="coerce")
df_p["lat"] = df_p["latitud"].map(lambda v: parse_coord(v, "lat"))
df_p["lon"] = df_p["longitud"].map(lambda v: parse_coord(v, "lon"))
df_p["edad"] = pd.to_numeric(df_p["edad"], errors="coerce")
edad_centinela = (df_p["edad"] < 0).sum()
df_p.loc[df_p["edad"] < 0, "edad"] = np.nan

categoricas = [
    "sexo", "estado_civil", "medio_transporte", "nivel_academico", "testigo",
    "conducta", "modalidad", "conducta_especial", "arma_medio", "nombre_barrio",
    "codigo_barrio", "codigo_comuna", "lugar", "sede_receptora", "bien",
    "categoria_bien", "color",
]
for col in categoricas:
    df_p[col] = recodificar_centinela(df_p[col])

na_parseo = pd.DataFrame({
    "na_lectura": na_antes,
    "na_despues_parseo": df_p.isna().sum(),
})
na_parseo["nacidos_en_parseo"] = na_parseo["na_despues_parseo"] - na_parseo["na_lectura"]
print(f"Fechas no parseadas: {df_p['fecha'].isna().sum()}")
print(f"Edades centinela (-1) convertidas a NaN: {edad_centinela}")
print(f"Coords fuera del valle de Aburrá: {((df_p['lat'] < 6.0) | (df_p['lat'] > 6.5) | (df_p['lon'] < -75.8) | (df_p['lon'] > -75.4)).sum()}")
na_parseo.loc[(na_parseo["nacidos_en_parseo"] != 0) | (na_parseo["na_despues_parseo"] > 0)]
```

## Depuración

Reglas aplicadas:

| Decisión | Justificación |
| --- | --- |
| Eliminar `nivel_academico` y `testigo` | 100 % `Sin dato`: no hay señal |
| Eliminar `conducta` | Constante (`Hurto a persona`) |
| Eliminar copias exactas | Error de captura; el mismo bien no ocurrió dos veces idéntico |
| **No** colapsar filas del mismo hecho | Un hurto puede reportar varios bienes; eso es información |
| Edad `< 0` | Imposible: ya es `NaN`, no se borra la fila |
| `color` y `conducta_especial` | > 60 % faltantes: no se imputan; se dejan fuera del análisis principal |
| Coordenadas atípicas | Tras el parseo, todas caen en Medellín: no se recortan como outliers |

```python
n0 = len(df_p)
df_d = df_p.drop_duplicates(keep="first").copy()
n_dups = n0 - len(df_d)

cols_vacias = [c for c in ["nivel_academico", "testigo", "conducta"] if c in df_d.columns]
df_d = df_d.drop(columns=cols_vacias)

# Identificador de hecho ANTES de imputar coordenadas
df_d["id_hecho"] = (
    df_d["fecha"].astype(str)
    + "|" + df_d["latitud"].astype(str)
    + "|" + df_d["longitud"].astype(str)
    + "|" + df_d["sexo"].astype(str)
    + "|" + df_d["edad"].astype(str)
    + "|" + df_d["nombre_barrio"].astype(str)
    + "|" + df_d["lugar"].astype(str)
    + "|" + df_d["modalidad"].astype(str)
)

print(f"Eliminadas {n_dups} filas duplicadas exactas ({n_dups / n0:.1%})")
print(f"Registros (bienes): {len(df_d)}")
print(f"Hechos únicos aproximados: {df_d['id_hecho'].nunique()}")
print(df_d["sexo"].value_counts(dropna=False))
print(df_d["edad"].describe())
```

## Imputación

- **Lat/lon (~4 %):** MAR plausible (falta más en algunos barrios). Se imputa con la **mediana del barrio**; si el barrio no tiene coords observadas, la de la **comuna**. No se usa la media global: mezclaría el centro con El Poblado.
- **Edad (centinelas, < 1 %):** muestreo empírico (hot-deck) **dentro del sexo**, con semilla fija, para no clavar un pico en la media.
- **Sexo, estado civil, color, conducta especial:** no se imputan. La moda inflaría una clase; en color/conducta especial el hueco es estructural.

Se compara la desviación de la edad antes y después: si cayera de golpe, el método estaría aplastando la distribución.

```python
rng = np.random.default_rng(42)
df_i = df_d.copy()
df_i["coord_imputada"] = df_i["lat"].isna()
df_i["edad_imputada"] = df_i["edad"].isna()

centro_barrio = df_i.groupby("nombre_barrio")[["lat", "lon"]].median()
centro_comuna = df_i.groupby("codigo_comuna")[["lat", "lon"]].median()

for idx in df_i.index[df_i["lat"].isna()]:
    barrio = df_i.at[idx, "nombre_barrio"]
    comuna = df_i.at[idx, "codigo_comuna"]
    if pd.notna(barrio) and barrio in centro_barrio.index and pd.notna(centro_barrio.loc[barrio, "lat"]):
        df_i.at[idx, "lat"] = centro_barrio.loc[barrio, "lat"]
        df_i.at[idx, "lon"] = centro_barrio.loc[barrio, "lon"]
    elif pd.notna(comuna) and comuna in centro_comuna.index and pd.notna(centro_comuna.loc[comuna, "lat"]):
        df_i.at[idx, "lat"] = centro_comuna.loc[comuna, "lat"]
        df_i.at[idx, "lon"] = centro_comuna.loc[comuna, "lon"]

edad_antes = df_d["edad"].describe()
sexo_grupo = df_i["sexo"].fillna("_sin_sexo")
for grupo, sub in df_i["edad"].groupby(sexo_grupo):
    faltan = sub.isna()
    n_faltan = int(faltan.sum())
    if n_faltan == 0:
        continue
    observados = sub.dropna()
    if observados.empty:
        observados = df_i["edad"].dropna()
    df_i.loc[sub.index[faltan], "edad"] = rng.choice(
        observados.to_numpy(), size=n_faltan, replace=True
    )

edad_despues = df_i["edad"].describe()
comparacion_edad = pd.concat([edad_antes, edad_despues], axis=1, keys=["antes", "despues"])
print("coordenadas aún faltantes:", df_i["lat"].isna().sum())
print("edades imputadas:", int(df_i["edad_imputada"].sum()))
print("coords imputadas:", int(df_i["coord_imputada"].sum()))
comparacion_edad
```

```python
COMUNAS = {
    "1": "Popular", "2": "Santa Cruz", "3": "Manrique", "4": "Aranjuez",
    "5": "Castilla", "6": "Doce de Octubre", "7": "Robledo", "8": "Villa Hermosa",
    "9": "Buenos Aires", "10": "La Candelaria", "11": "Laureles-Estadio",
    "12": "La América", "13": "San Javier", "14": "El Poblado", "15": "Guayabal",
    "16": "Belén", "50": "Palmitas", "60": "San Antonio de Prado",
    "70": "Altavista", "80": "San Cristóbal", "90": "Santa Elena",
}
DIAS = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo"]

df_i["comuna"] = df_i["codigo_comuna"].map(COMUNAS).fillna("Sin dato")
df_i["anio"] = df_i["fecha"].dt.year
df_i["mes"] = df_i["fecha"].dt.month
df_i["hora"] = df_i["fecha"].dt.hour
df_i["dia"] = pd.Categorical(df_i["fecha"].dt.dayofweek.map(dict(enumerate(DIAS))), categories=DIAS, ordered=True)
df_i["anio_mes"] = df_i["fecha"].dt.to_period("M").dt.to_timestamp()
df_i["franja"] = pd.cut(
    df_i["hora"],
    bins=[-0.1, 5, 11, 17, 23],
    labels=["Madrugada (0-5)", "Mañana (6-11)", "Tarde (12-17)", "Noche (18-23)"],
)

hechos = df_i.drop_duplicates("id_hecho").copy()
print(hechos.shape, "hechos únicos")

from pathlib import Path

raiz = Path.cwd()
salida = raiz / "data" / "processed" if (raiz / "data").exists() else raiz.parent / "data" / "processed"
salida.mkdir(parents=True, exist_ok=True)
df_i.to_csv(salida / "hurtos_medellin_bienes.csv", index=False)
hechos.to_csv(salida / "hurtos_medellin_hechos.csv", index=False)
print("guardado en", salida.resolve())
```

## Gráficos descriptivos

Los conteos de **hechos** (no de bienes) evitan inflar un hurto con varios objetos. Las recurrencias que buscamos: tiempo, hora, lugar y modus operandi.

```python
fig, axes = plt.subplots(2, 1, figsize=(11, 8), sharex=False)

mensual = hechos.groupby("anio_mes").size()
axes[0].plot(mensual.index, mensual.values, color="#1f77b4", lw=1.4)
axes[0].set_title("Hechos por mes (2004-2020)")
axes[0].set_ylabel("Hechos")

anual = hechos.groupby("anio").size()
axes[1].bar(anual.index.astype(int), anual.values, color="#1f77b4")
axes[1].set_title("Hechos por año")
axes[1].set_xlabel("Año")
axes[1].set_ylabel("Hechos")
fig.tight_layout()
plt.show()
```

```python
pivot = (
    hechos.pivot_table(index="dia", columns="hora", values="id_hecho", aggfunc="count")
    .reindex(DIAS)
    .fillna(0)
)

fig, ax = plt.subplots(figsize=(12, 4.2))
sns.heatmap(pivot, cmap="YlOrRd", ax=ax, linewidths=0.2)
ax.set_title("Recurrencia temporal: día × hora")
ax.set_xlabel("Hora del día")
ax.set_ylabel("")
fig.tight_layout()
plt.show()

print(hechos["franja"].value_counts(normalize=True).mul(100).round(1))
```

La franja más cargada es la **mañana (6-11)**, con **cosquilleo** dominante (hora pico de bus/metro). En **noche y madrugada** sube el peso del **atraco**. El fenómeno no está repartido de forma uniforme en el día.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

mod = hechos["modalidad"].value_counts().head(8)
sns.barplot(x=mod.values, y=mod.index, ax=axes[0], color="#2ca02c")
axes[0].set_title("Modalidad del hurto")
axes[0].set_xlabel("Hechos")

arma = hechos["arma_medio"].fillna("Sin dato").value_counts()
sns.barplot(x=arma.values, y=arma.index, ax=axes[1], color="#9467bd")
axes[1].set_title("Medio / arma")
axes[1].set_xlabel("Hechos")
fig.tight_layout()
plt.show()

mix = pd.crosstab(hechos["franja"], hechos["modalidad"], normalize="index") * 100
print(mix.reindex(columns=["Cosquilleo", "Atraco"]).round(1))
```

### Lectura breve

1. **Recurrencia espacial:** el centro (Candelaria) y nodos de transporte concentran el fenómeno en este dataset.
2. **Recurrencia horaria:** pico en la mañana (cosquilleo en transporte); atraco relativamente más frecuente de noche.
3. **Dos modus operandi:** cosquilleo en el sistema de transporte vs atraco armado.
4. **Objetivo típico:** celular y dinero, víctimas adultas jóvenes.
5. **Limitación:** el archivo está filtrado a bus/taxi/metro. No se puede inferir la geografía completa del hurto en Medellín.

`df_i` = registros de bienes limpios e imputados. `hechos` = un registro por evento.
