```python
import pandas as pd

RUTA = "../data/WB_RISE.csv"

# Vista rápida: 5 filas, sin cargar el archivo completo
muestra = pd.read_csv(RUTA, nrows=5)
print(muestra.columns.tolist())
print(muestra.dtypes)
muestra
```

```python
df = pd.read_csv(
    RUTA,
    sep=",",
    encoding="utf-8",
    decimal=".",
    na_values=["", "NA", "N/A", "-", "null", "999"],
    skipinitialspace=True,
)

print(df.shape)
print(df.info())
df.head()
```

```python
import kagglehub
from kagglehub import KaggleDatasetAdapter

file_path = "robbery of people in Medellin.csv"

df = kagglehub.dataset_load(
    KaggleDatasetAdapter.PANDAS,
    "darwinberrio/robbery-to-people-in-medellincolombia",
    file_path,
    pandas_kwargs={
        "sep": ";",
        "encoding": "utf-8-sig",
        "compression": "zip",
    },
)

print(df.shape)
df.head()
```
